# CORDEX Catalog: add 2 columns

Adding two new columns in the CORDEX intake with a separate 'region' (ARC, EUR...) and 'horizontal_resolution' (44, 11...).

date: 2026-01-07


```python
import intake
import pandas as pd
import re
```

```python
cat_url = "https://raw.githubusercontent.com/cp4cds/c3s_34g_manifests/master/intake/catalogs/c3s.yaml"

cat = intake.open_catalog(cat_url)
list(cat)
```

```python
df_cordex = cat['c3s-cordex'].read()
df_cordex.head()
```

```python
df_cordex.nunique()
```

## Add two new columns derived from domain

domain = {region}-{horizontal_resolution}

```python
df_cordex[["region", "horizontal_resolution"]] = (
    df_cordex["domain"]
    .astype("string")
    .str.split("-", n=1, expand=True)
)
df_cordex.head()
```

```python
df_cordex.nunique()
```

```python
list(df_cordex.domain.unique())
```

```python
list(df_cordex.region.unique())
```

```python
list(df_cordex.horizontal_resolution.unique())
```

## Write new catalog

```python
from datetime import datetime, UTC

last_updated = datetime.now(UTC)
version = last_updated.strftime('v%Y%m%d')
cat_name = f"c3s-cordex_{version}.csv.gz"
cat_path = f"../intake/catalogs/c3s-cordex/{cat_name}"

df_cordex.to_csv(cat_path, index=False, compression="gzip")
```
