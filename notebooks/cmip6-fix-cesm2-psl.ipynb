````python
import intake
import pandas as pd
````

````python
cat_url = "https://raw.githubusercontent.com/cp4cds/c3s_34g_manifests/master/intake/catalogs/c3s.yaml"
#cat_url = "https://github.com/cp4cds/c3s_34g_manifests/raw/add-fixes-for-cmip6-cordex/intake/catalogs/c3s.yaml"

cat = intake.open_catalog(cat_url)
list(cat)
````

## Load CMIP6 catalog

````python
df_cmip6 = cat['c3s-cmip6'].read()
df_cmip6
````

````python
df_cmip6.nunique()
````

## Drop duplicate netcdf files covering the full time period in cesm2 psl

````python
group_cols = [
    "mip_era",
    "activity_id",
    "institution_id",
    "source_id",
    "experiment_id",
    "member_id",
    "table_id",
    "variable_id",
    "grid_label",
    "version",
]
````

````python
def drop_full_range_files(df, group_cols):
    to_drop = []

    for group_values, g in df.groupby(group_cols):
        # Print the current group being checked
        group_info = dict(zip(group_cols, group_values))
        print(f"Checking dataset group: {group_info}")

        if len(g) <= 1:
            print("  Only one file in group → nothing to drop")
            continue

        min_start = g.start_time.min()
        max_end   = g.end_time.max()

        mask = (g.start_time == min_start) & (g.end_time == max_end)

        if mask.sum() == 1:
            for path in g.loc[mask, "path"]:
                print(f"  Dropping full-range file: {path}")
            to_drop.extend(g.loc[mask].index)
        else:
            print("  Group is clean → no full-range file to drop")

    print(f"Total files dropped: {len(to_drop)}")
    return df.drop(index=to_drop)

````

````python
df_psl = df_cmip6.loc[
    (df_cmip6.source_id == "CESM2") &
    (df_cmip6.variable_id == "psl")
]

df_psl.nunique()
````

````python
df_psl_clean = drop_full_range_files(df_psl, group_cols)
df_psl_clean.nunique()
````

## Update cmip6 catalog

````python
# df_psl: original CESM2 psl subset
# df_psl_clean: after removing full-range files

dropped_files = df_psl.index.difference(df_psl_clean.index)
# or, for paths:
dropped_paths = df_psl.loc[dropped_files, "path"].tolist()

print("Dropped files:")
for p in dropped_paths:
    print(f"  {p}")
````

````python
df_cmip6_clean = df_cmip6.drop(index=dropped_files)

df_cmip6_clean.nunique()
````

````python
removed_files = df_cmip6.index.difference(df_cmip6_clean.index)
print("Files removed from catalog:")
print(df_cmip6.loc[removed_files, ["path", "start_time", "end_time"]])
````

## Write fixed catalog

````python
from datetime import datetime, UTC

last_updated = datetime.now(UTC)
version = last_updated.strftime('v%Y%m%d')
cat_name = f"c3s-cmip6_{version}.csv.gz"
cat_path = f"../intake/catalogs/c3s-cmip6/{cat_name}"

df_cmip6_clean.to_csv(cat_path, index=False, compression="gzip")
````
