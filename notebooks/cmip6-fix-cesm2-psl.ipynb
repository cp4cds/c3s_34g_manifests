````python
import intake
import pandas as pd
````

````python
cat_url = "https://raw.githubusercontent.com/cp4cds/c3s_34g_manifests/master/intake/catalogs/c3s.yaml"
#cat_url = "https://github.com/cp4cds/c3s_34g_manifests/raw/add-fixes-for-cmip6-cordex/intake/catalogs/c3s.yaml"

cat = intake.open_catalog(cat_url)
list(cat)
````

## Load CMIP6 catalog

````python
df_cmip6 = cat['c3s-cmip6'].read()
df_cmip6
````

````python
df_cmip6.nunique()
````

````python
group_cols = [
    "mip_era",
    "activity_id",
    "institution_id",
    "source_id",
    "experiment_id",
    "member_id",
    "table_id",
    "variable_id",
    "grid_label",
    "version",
]
````

````python
def drop_full_range_files(df, group_cols):
    to_drop = []

    for _, g in df.groupby(group_cols):
        if len(g) <= 1:
            continue

        min_start = g.start_time.min()
        max_end   = g.end_time.max()

        mask = (g.start_time == min_start) & (g.end_time == max_end)

        # Drop only if exactly one file spans the full range
        if mask.sum() == 1:
            to_drop.extend(g.loc[mask].index)

    return df.drop(index=to_drop)
````

````python
df_psl = df_cmip6.loc[
    (df_cmip6.source_id == "CESM2") &
    (df_cmip6.variable_id == "psl")
]

df_new = drop_full_range_files(df_psl, group_cols)
df_new
````

````python
df_new.nunique()
````

## Write fixed catalog

````python
from datetime import datetime, UTC

last_updated = datetime.now(UTC)
version = last_updated.strftime('v%Y%m%d')
cat_name = f"c3s-cmip6_{version}.csv.gz"
cat_path = f"../intake/catalogs/c3s-cmip6/{cat_name}"

df_new.to_csv(cat_path, index=False, compression="gzip")
````
